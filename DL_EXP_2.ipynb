```python
# EXPERIMENT NO: 2
# COMPARATIVE STUDY OF ACTIVATION FUNCTIONS AND
# OPTIMIZATION ALGORITHMS IN DEEP LEARNING
# TASK A: VISUALIZATION OF ACTIVATION FUNCTIONS
# Roll No: 24BAD103

import numpy as np
import matplotlib.pyplot as plt
import time

# Define Activation Functions

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

# Generate Input Values

x = np.linspace(-10, 10, 1000)

sigmoid_output = sigmoid(x)
tanh_output = tanh(x)
relu_output = relu(x)

# Plot Activation Functions

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.plot(x, sigmoid_output)
plt.title("Sigmoid Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)

plt.subplot(1, 3, 2)
plt.plot(x, tanh_output)
plt.title("Tanh Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)

plt.subplot(1, 3, 3)
plt.plot(x, relu_output)
plt.title("ReLU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)

plt.suptitle("Activation Functions - Roll No: 24BAD103")
plt.tight_layout()
plt.show()

# Calculate Gradients

sigmoid_gradient = sigmoid_output * (1 - sigmoid_output)

tanh_gradient = 1 - tanh_output ** 2

relu_gradient = np.where(x > 0, 1, 0)

# Plot Gradient Behavior

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.plot(x, sigmoid_gradient)
plt.title("Sigmoid Gradient")
plt.xlabel("Input (x)")
plt.ylabel("Gradient")
plt.grid(True)

plt.subplot(1, 3, 2)
plt.plot(x, tanh_gradient)
plt.title("Tanh Gradient")
plt.xlabel("Input (x)")
plt.ylabel("Gradient")
plt.grid(True)

plt.subplot(1, 3, 3)
plt.plot(x, relu_gradient)
plt.title("ReLU Gradient")
plt.xlabel("Input (x)")
plt.ylabel("Gradient")
plt.grid(True)

plt.suptitle("Gradient Behavior of Activation Functions")
plt.tight_layout()
plt.show()

# Computational Efficiency

large_x = np.random.randn(1000000)

start = time.time()

for _ in range(10):
    sigmoid(large_x)

sigmoid_time = time.time() - start

start = time.time()

for _ in range(10):
    tanh(large_x)

tanh_time = time.time() - start

start = time.time()

for _ in range(10):
    relu(large_x)

relu_time = time.time() - start

print("\nCOMPUTATIONAL EFFICIENCY")


print(f"Sigmoid : {sigmoid_time:.6f} seconds")
print(f"Tanh    : {tanh_time:.6f} seconds")
print(f"ReLU    : {relu_time:.6f} seconds")

# Required Comparison

print("\nCOMPARISON OF ACTIVATION FUNCTIONS")


print("\n1. OUTPUT RANGE")
print("Sigmoid : 0 to 1")
print("Tanh    : -1 to +1")
print("ReLU    : 0 to infinity")

print("\n2. SATURATION REGIONS")
print("Sigmoid : Saturates for large positive and negative inputs.")
print("Tanh    : Saturates for large positive and negative inputs.")
print("ReLU    : Output is 0 for negative inputs; no saturation for positive inputs.")

print("\n3. GRADIENT BEHAVIOR")
print("Sigmoid : Gradient becomes very small in saturation regions.")
print("Tanh    : Gradient becomes very small in saturation regions.")
print("ReLU    : Gradient is 0 for negative inputs and 1 for positive inputs.")

print("\n4. COMPUTATIONAL EFFICIENCY")
print("Sigmoid : Moderate computational cost.")
print("Tanh    : Moderate computational cost.")
print("ReLU    : Computationally efficient and simple.")

print("\n5. TYPICAL APPLICATIONS")
print("Sigmoid : Binary classification output layers.")
print("Tanh    : Hidden layers and recurrent neural networks.")
print("ReLU    : Hidden layers of deep neural networks.")

print("\nTASK A COMPLETED SUCCESSFULLY")
```

```python
# EXPERIMENT NO: 2
# TASK B: PERFORMANCE COMPARISON OF ACTIVATION FUNCTIONS
# Roll No: 24BAD103

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize pixel values
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Flatten images
x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)

# Convert labels to categorical format
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# Use a validation set from the training data
x_val = x_train[-10000:]
y_val = y_train[-10000:]

x_train_new = x_train[:-10000]
y_train_new = y_train[:-10000]

print("Training samples  :", x_train_new.shape[0])
print("Validation samples:", x_val.shape[0])
print("Testing samples   :", x_test.shape[0])


# Function to create identical ANN architectures
def create_model(activation):

    model = Sequential([
        Dense(128, activation=activation, input_shape=(784,)),
        Dense(64, activation=activation),
        Dense(10, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# Train ANN with Sigmoid activation
print("\nTraining with Sigmoid activation...")

sigmoid_model = create_model("sigmoid")

sigmoid_history = sigmoid_model.fit(
    x_train_new,
    y_train_new,
    validation_data=(x_val, y_val),
    epochs=20,
    batch_size=128,
    verbose=1
)


# Train ANN with Tanh activation
print("\nTraining with Tanh activation...")

tanh_model = create_model("tanh")

tanh_history = tanh_model.fit(
    x_train_new,
    y_train_new,
    validation_data=(x_val, y_val),
    epochs=20,
    batch_size=128,
    verbose=1
)


# Train ANN with ReLU activation
print("\nTraining with ReLU activation...")

relu_model = create_model("relu")

relu_history = relu_model.fit(
    x_train_new,
    y_train_new,
    validation_data=(x_val, y_val),
    epochs=20,
    batch_size=128,
    verbose=1
)


# Evaluate models on test data
sigmoid_loss, sigmoid_accuracy = sigmoid_model.evaluate(
    x_test, y_test, verbose=0
)

tanh_loss, tanh_accuracy = tanh_model.evaluate(
    x_test, y_test, verbose=0
)

relu_loss, relu_accuracy = relu_model.evaluate(
    x_test, y_test, verbose=0
)


# Display final results
print("\n" + "=" * 70)
print("FINAL MODEL PERFORMANCE")
print("=" * 70)

print(f"\nSigmoid:")
print(f"Training Accuracy   : {sigmoid_history.history['accuracy'][-1]:.4f}")
print(f"Validation Accuracy : {sigmoid_history.history['val_accuracy'][-1]:.4f}")
print(f"Training Loss       : {sigmoid_history.history['loss'][-1]:.4f}")
print(f"Validation Loss     : {sigmoid_history.history['val_loss'][-1]:.4f}")
print(f"Test Accuracy       : {sigmoid_accuracy:.4f}")

print(f"\nTanh:")
print(f"Training Accuracy   : {tanh_history.history['accuracy'][-1]:.4f}")
print(f"Validation Accuracy : {tanh_history.history['val_accuracy'][-1]:.4f}")
print(f"Training Loss       : {tanh_history.history['loss'][-1]:.4f}")
print(f"Validation Loss     : {tanh_history.history['val_loss'][-1]:.4f}")
print(f"Test Accuracy       : {tanh_accuracy:.4f}")

print(f"\nReLU:")
print(f"Training Accuracy   : {relu_history.history['accuracy'][-1]:.4f}")
print(f"Validation Accuracy : {relu_history.history['val_accuracy'][-1]:.4f}")
print(f"Training Loss       : {relu_history.history['loss'][-1]:.4f}")
print(f"Validation Loss     : {relu_history.history['val_loss'][-1]:.4f}")
print(f"Test Accuracy       : {relu_accuracy:.4f}")


# Plot Training and Validation Accuracy
plt.figure(figsize=(10, 6))

plt.plot(
    sigmoid_history.history["accuracy"],
    label="Sigmoid Training"
)

plt.plot(
    sigmoid_history.history["val_accuracy"],
    label="Sigmoid Validation"
)

plt.plot(
    tanh_history.history["accuracy"],
    label="Tanh Training"
)

plt.plot(
    tanh_history.history["val_accuracy"],
    label="Tanh Validation"
)

plt.plot(
    relu_history.history["accuracy"],
    label="ReLU Training"
)

plt.plot(
    relu_history.history["val_accuracy"],
    label="ReLU Validation"
)

plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()


# Plot Training and Validation Loss
plt.figure(figsize=(10, 6))

plt.plot(
    sigmoid_history.history["loss"],
    label="Sigmoid Training"
)

plt.plot(
    sigmoid_history.history["val_loss"],
    label="Sigmoid Validation"
)

plt.plot(
    tanh_history.history["loss"],
    label="Tanh Training"
)

plt.plot(
    tanh_history.history["val_loss"],
    label="Tanh Validation"
)

plt.plot(
    relu_history.history["loss"],
    label="ReLU Training"
)

plt.plot(
    relu_history.history["val_loss"],
    label="ReLU Validation"
)

plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()


# Compare convergence
print("\n" + "=" * 70)
print("CONVERGENCE COMPARISON")
print("=" * 70)

print(f"Sigmoid epochs : {len(sigmoid_history.history['loss'])}")
print(f"Tanh epochs    : {len(tanh_history.history['loss'])}")
print(f"ReLU epochs    : {len(relu_history.history['loss'])}")

print("\n" + "=" * 70)
print("TASK B COMPLETED SUCCESSFULLY")
print("=" * 70)

```

```python
# EXPERIMENT NO: 2
# TASK C: COMPARISON OF OPTIMIZATION ALGORITHMS
# Roll No: 24BAD103

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import SGD, RMSprop, Adam

# Set random seeds
np.random.seed(42)
tf.random.set_seed(42)

# Create the same ANN architecture for all optimizers
def create_model(optimizer):

    model = Sequential([
        Dense(128, activation="relu", input_shape=(784,)),
        Dense(64, activation="relu"),
        Dense(10, activation="softmax")
    ])

    model.compile(
        optimizer=optimizer,
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# Define optimizers

sgd = SGD(learning_rate=0.01)

momentum = SGD(
    learning_rate=0.01,
    momentum=0.9
)

rmsprop = RMSprop(
    learning_rate=0.001
)

adam = Adam(
    learning_rate=0.001
)


# Train with SGD

print("Training with SGD...")

sgd_model = create_model(sgd)

sgd_history = sgd_model.fit(
    x_train_new,
    y_train_new,
    validation_data=(x_val, y_val),
    epochs=20,
    batch_size=128,
    verbose=1
)


# Train with Momentum

print("\nTraining with Momentum...")

momentum_model = create_model(momentum)

momentum_history = momentum_model.fit(
    x_train_new,
    y_train_new,
    validation_data=(x_val, y_val),
    epochs=20,
    batch_size=128,
    verbose=1
)


# Train with RMSProp

print("\nTraining with RMSProp...")

rmsprop_model = create_model(rmsprop)

rmsprop_history = rmsprop_model.fit(
    x_train_new,
    y_train_new,
    validation_data=(x_val, y_val),
    epochs=20,
    batch_size=128,
    verbose=1
)


# Train with Adam

print("\nTraining with Adam...")

adam_model = create_model(adam)

adam_history = adam_model.fit(
    x_train_new,
    y_train_new,
    validation_data=(x_val, y_val),
    epochs=20,
    batch_size=128,
    verbose=1
)


# Evaluate all models

sgd_loss, sgd_accuracy = sgd_model.evaluate(
    x_test, y_test, verbose=0
)

momentum_loss, momentum_accuracy = momentum_model.evaluate(
    x_test, y_test, verbose=0
)

rmsprop_loss, rmsprop_accuracy = rmsprop_model.evaluate(
    x_test, y_test, verbose=0
)

adam_loss, adam_accuracy = adam_model.evaluate(
    x_test, y_test, verbose=0
)


# Display final results

print("\n" + "=" * 70)
print("FINAL PERFORMANCE COMPARISON")
print("=" * 70)

print("\nSGD")
print(f"Training Loss     : {sgd_history.history['loss'][-1]:.4f}")
print(f"Validation Loss   : {sgd_history.history['val_loss'][-1]:.4f}")
print(f"Training Accuracy : {sgd_history.history['accuracy'][-1]:.4f}")
print(f"Validation Accuracy: {sgd_history.history['val_accuracy'][-1]:.4f}")

print("\nMomentum")
print(f"Training Loss     : {momentum_history.history['loss'][-1]:.4f}")
print(f"Validation Loss   : {momentum_history.history['val_loss'][-1]:.4f}")
print(f"Training Accuracy : {momentum_history.history['accuracy'][-1]:.4f}")
print(f"Validation Accuracy: {momentum_history.history['val_accuracy'][-1]:.4f}")

print("\nRMSProp")
print(f"Training Loss     : {rmsprop_history.history['loss'][-1]:.4f}")
print(f"Validation Loss   : {rmsprop_history.history['val_loss'][-1]:.4f}")
print(f"Training Accuracy : {rmsprop_history.history['accuracy'][-1]:.4f}")
print(f"Validation Accuracy: {rmsprop_history.history['val_accuracy'][-1]:.4f}")

print("\nAdam")
print(f"Training Loss     : {adam_history.history['loss'][-1]:.4f}")
print(f"Validation Loss   : {adam_history.history['val_loss'][-1]:.4f}")
print(f"Training Accuracy : {adam_history.history['accuracy'][-1]:.4f}")
print(f"Validation Accuracy: {adam_history.history['val_accuracy'][-1]:.4f}")


# Loss vs Epoch

plt.figure(figsize=(10, 6))

plt.plot(sgd_history.history["loss"], label="SGD")
plt.plot(momentum_history.history["loss"], label="Momentum")
plt.plot(rmsprop_history.history["loss"], label="RMSProp")
plt.plot(adam_history.history["loss"], label="Adam")

plt.title("Training Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training Loss")
plt.legend()
plt.grid(True)

plt.show()


# Validation Loss vs Epoch

plt.figure(figsize=(10, 6))

plt.plot(sgd_history.history["val_loss"], label="SGD")
plt.plot(momentum_history.history["val_loss"], label="Momentum")
plt.plot(rmsprop_history.history["val_loss"], label="RMSProp")
plt.plot(adam_history.history["val_loss"], label="Adam")

plt.title("Validation Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid(True)

plt.show()


# Accuracy vs Epoch

plt.figure(figsize=(10, 6))

plt.plot(sgd_history.history["accuracy"], label="SGD")
plt.plot(momentum_history.history["accuracy"], label="Momentum")
plt.plot(rmsprop_history.history["accuracy"], label="RMSProp")
plt.plot(adam_history.history["accuracy"], label="Adam")

plt.title("Training Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()


# Validation Accuracy vs Epoch

plt.figure(figsize=(10, 6))

plt.plot(sgd_history.history["val_accuracy"], label="SGD")
plt.plot(momentum_history.history["val_accuracy"], label="Momentum")
plt.plot(rmsprop_history.history["val_accuracy"], label="RMSProp")
plt.plot(adam_history.history["val_accuracy"], label="Adam")

plt.title("Validation Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()


# Convergence comparison

print("\n" + "=" * 70)
print("CONVERGENCE SPEED")
print("=" * 70)

print(f"SGD      : {len(sgd_history.history['loss'])} epochs")
print(f"Momentum : {len(momentum_history.history['loss'])} epochs")
print(f"RMSProp  : {len(rmsprop_history.history['loss'])} epochs")
print(f"Adam     : {len(adam_history.history['loss'])} epochs")


# Find best validation accuracy

results = {
    "SGD": max(sgd_history.history["val_accuracy"]),
    "Momentum": max(momentum_history.history["val_accuracy"]),
    "RMSProp": max(rmsprop_history.history["val_accuracy"]),
    "Adam": max(adam_history.history["val_accuracy"])
}

best_optimizer = max(results, key=results.get)

print("\n" + "=" * 70)
print("BEST OPTIMIZER")
print("=" * 70)

print(f"Best optimizer based on validation accuracy: {best_optimizer}")
print(f"Best validation accuracy: {results[best_optimizer]:.4f}")

print("\nTASK C COMPLETED SUCCESSFULLY")
```
